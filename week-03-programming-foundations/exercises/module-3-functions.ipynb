## 3. `return` sends a result back
The returned value can be stored and reused.

```python
def double_score(score):
    result = score * 2
    return result

new_score = double_score(7)
print(new_score)
```

## 4. Build an average function together
Follow the accumulator pattern, then return the result.

```python
def calculate_average(scores):
    total = 0
    for score in scores:
        total = total + score
    average = total / len(scores)
    return average

example = calculate_average([6, 8, 7, 9])
print(example)
```

**Checkpoint:** The output should be `7.5`. Test `[10, 10, 10]`.
