## 2. A `for` loop visits every item
Point to the loop variable. How many times will `print` run?

```python
scores = [6, 8, 7, 9]
for score in scores:
    print("Current score:", score)
```

## 3. An accumulator remembers a running result
Trace `score` and `total` on paper before running.

```python
scores = [6, 8, 7, 9]
total = 0
for score in scores:
    total = total + score
    print("score =", score, "total =", total)
print("Final total:", total)
```

**Check your trace:** The totals should be `6`, `14`, `21`, and `30`.

## 4. A conditional can filter inside a loop
Predict which scores will print.

```python
for score in scores:
    if score >= 8:
        print(score)
```

**Try it:** Change the condition so only scores below 8 print.

## 5. A `while` loop repeats while a condition stays true
Point to the line that changes `countdown`. Without it, the loop would never stop.

```python
countdown = 3
while countdown > 0:
    print(countdown)
    countdown = countdown - 1
print("Go!")
```

**Checkpoint:** Start at 5. Then change the loop so it stops after printing 2.
