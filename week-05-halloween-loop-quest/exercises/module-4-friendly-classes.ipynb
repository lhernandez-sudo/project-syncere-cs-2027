## Why would an engineer use a class?

A game may have ten players, fifty enemies, and hundreds of power-ups. Separate variables such as `enemy_1_health`, `enemy_2_health`, and `enemy_3_health` become difficult to manage.

A **class** is a blueprint that groups related data and behavior. An **object** is one thing built from that blueprint.

| Everyday idea | Python idea | Game example |
|---|---|---|
| Blueprint | Class | `Player` |
| One thing built from it | Object | `mario` or `luigi` |
| Facts about the thing | Attributes | `name`, `health`, `coins` |
| Actions it can perform | Methods | `jump()`, `take_damage()` |

Classes are useful when several things share the same kinds of data and actions but need their **own state**.


## Theory in one picture

```text
Player class (blueprint)
├── attributes: name, health, coins
└── methods: collect_coin(), take_damage()

mario object                 luigi object
name = "Mario"               name = "Luigi"
health = 3                   health = 3
coins = 5                    coins = 0
```

Changing Mario's coins should not change Luigi's coins. Each object owns its own attribute values.


````python
class Player:
    def __init__(self, name):
        self.name = name
        self.health = 3
        self.coins = 0

mario = Player("Mario")
luigi = Player("Luigi")
mario.coins = 5

print(mario.name, mario.health, mario.coins)
print(luigi.name, luigi.health, luigi.coins)
integrity_check("Independent objects", [
    (mario is not luigi, "Mario and Luigi must be different objects."),
    (mario.coins == 5, "Mario should own five coins."),
    (luigi.coins == 0, "Changing Mario must not change Luigi."),
])
````

## What do `__init__` and `self` mean?

- `__init__` runs automatically when a new object is created.
- `self` means **this specific object**.
- `self.name = name` stores the incoming `name` on this object.
- `Player("Peach")` calls `__init__` and creates one new `Player` object.

Read `self.health = self.health - amount` as: **this object's health becomes its old health minus the amount**.

## Applicability — why not use only a dictionary?

A dictionary can store one character's data, but a class keeps the rules next to the data.

```python
# Dictionary: code elsewhere must remember every rule
enemy = {"name": "Goomba", "health": 2}

# Class: the object knows how damage works
enemy = Enemy("Goomba", 2)
enemy.take_damage(1)
```

Classes are not automatically better for every problem. Use a class when you need several related objects with shared behaviors and independent state.

## Objects work with concepts you already know

Classes do not replace variables, conditions, loops, lists, or functions. They organize them.

The next cell stores objects in a list, loops over them, calls a method, and uses a condition.


````python
enemies = [Enemy("Goomba", 2), Enemy("Koopa", 3), Enemy("Boo", 1)]
defeated_names = []

for enemy in enemies:
    enemy.take_damage(2)
    if enemy.is_defeated():
        defeated_names.append(enemy.name)

print("Defeated:", defeated_names)
integrity_check("Object-loop integrity", [
    (defeated_names == ["Goomba", "Boo"], "Use each object's method and current state."),
    (enemies[1].health == 1, "Koopa should survive with one health."),
])
````

## Friendly OOP summary

- A **class** defines a reusable blueprint.
- An **object** is one instance of that class.
- An **attribute** stores object state.
- A **method** defines an action the object can perform.
- `__init__` sets the starting state.
- `self` refers to the specific object receiving the method call.
