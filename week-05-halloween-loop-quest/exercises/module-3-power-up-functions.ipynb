## Level 2 — Return sends a value back

`print()` shows a value. `return` gives a value back to the calling code so it can be saved or reused.


```python
def add_bonus(score, bonus):
    new_score = score + bonus
    return new_score

final_score = add_bonus(10, 5)
print("Final:", final_score)
checkpoint("Return value", final_score == 15)
```

### Predict three parts

1. Arguments: `7` and `3`  
2. Parameters: `damage` and `armor`  
3. Returned value: ?


```python
def hit_points(damage, armor):
    result = damage - armor
    return result

points_lost = hit_points(7, 3)
print(points_lost)
checkpoint("Function trace", points_lost == 4)
```

## Level 3 — Functions can contain decisions


```python
def rank_player(score):
    if score >= 20:
        return "GOLD"
    elif score >= 10:
        return "SILVER"
    else:
        return "BRONZE"

print(rank_player(8))
print(rank_player(14))
print(rank_player(25))
```

## Level 4 — Functions can contain loops

Trace `total` after every score before running.


```python
def total_scores(scores):
    total = 0
    for score in scores:
        total += score
    return total

team_total = total_scores([4, 6, 5])
print(team_total)
checkpoint("Function loop", team_total == 15)
```
