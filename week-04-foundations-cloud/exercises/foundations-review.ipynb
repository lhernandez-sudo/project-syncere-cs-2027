```python
def project_average(scores: list[int]) -> float:
    total = 0
    for score in scores:
        # TODO A: Add score to total.
        total = total
    return total / len(scores)

print(project_average([6, 8, 7]))  # Expected: 7.0
```

### Project checkpoint B: feedback

```python
def project_message(average: float) -> str:
    if average >= 8:
        return "Strong foundation. Explain one concept to someone else."
    # TODO B: Add a middle branch for averages at least 5.
    return "Choose one concept and trace another example."

print(project_message(8))
print(project_message(6))
print(project_message(3))
```
